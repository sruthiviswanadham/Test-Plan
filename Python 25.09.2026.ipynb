```python
def max_sum(arr):
    current = arr[0]
    maximum = arr[0]

    for i in range(1, len(arr)):
        current = max(arr[i], current + arr[i])
        maximum = max(maximum, current)

    return maximum


arr = [2,4,5,6,-1]
print(max_sum(arr))
```

```python
def trap(height):
    n = len(height)
    water = 0

    for i in range(n):
        left_max = max(height[:i + 1])
        right_max = max(height[i:])

        trapped = min(left_max, right_max) - height[i]

        water += trapped

    return water


height = [1,2,4,-3,4]

print(trap(height))
```

```python
def max_performance(arr):
    current = arr[0]
    maximum = arr[0]

    for i in range(1, len(arr)):
        current = max(arr[i], current + arr[i])
        maximum = max(maximum, current)

    return maximum


arr = [-2,-5,4,1,14]
print(max_performance(arr))
```

```python
def max_product(arr):
    maximum = arr[0]
    minimum = arr[0]
    answer = arr[0]

    for x in arr[1:]:
        if x < 0:
            maximum, minimum = minimum, maximum

        maximum = max(x, maximum * x)
        minimum = min(x, minimum * x)

        answer = max(answer, maximum)

    return answer


arr = [-2,3,-4]

print(max_product(arr))
print(max_performance(arr))
```

```python
def longest_unique(arr):
    seen = set()
    left = 0
    max_len = 0

    for right in range(len(arr)):
        while arr[right] in seen:
            seen.remove(arr[left])
            left += 1

        seen.add(arr[right])

        max_len = max(max_len, right - left + 1)

    return max_len


arr = [1,2,3,1,2]

print(longest_unique(arr))
```

```python
def count_subarrays(arr, target):
    prefix = 0
    count = 0

    freq = {0: 1}

    for x in arr:
        prefix += x

        if prefix - target in freq:
            count += freq[prefix - target]

        freq[prefix] = freq.get(prefix, 0) + 1

    return count


arr = [1,2,3,-2,2]
target = 3

print(count_subarrays(arr, target))
```

```python
def group_skills(arr):
    groups = {}

    for word in arr:
        key = ''.join(sorted(word))

        if key not in groups:
            groups[key] = []

        groups[key].append(word)

    return list(groups.values())


arr = input().split()

print(group_skills(arr))
```

```python
def longest_consecutive(arr):
    nums = set(arr)
    maximum = 0

    for x in nums:

        if x - 1 not in nums:
            current = x
            length = 1

            while current + 1 in nums:
                current += 1
                length += 1

            maximum = max(maximum, length)

    return maximum


arr = [100,400,3,2,1]

print(longest_consecutive(arr))
```

```python
def merge_intervals(intervals):
    intervals.sort()

    result = []

    for interval in intervals:

        if not result or result[-1][1] < interval[0]:
            result.append(interval)

        else:
            result[-1][1] = max(result[-1][1], interval[1])

    return result


n = int(input())

intervals = []

for i in range(n):
    start, end = map(int, input().split())
    intervals.append([start, end])

print(merge_intervals(intervals))
```
